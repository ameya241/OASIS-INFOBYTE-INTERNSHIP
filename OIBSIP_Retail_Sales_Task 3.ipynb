# Missing data handling
#### **Age:** In age column there is numerical records so i used median. median is less affected than mean.
#### **Fare:** In Fare column there is numerical records so i used median and may contain extreme ticket prices.
#### **Embarked:** mode is used because it is categorical column and only a small number of values are missing.
#### **Cabin,Boat,Body,Home.dest:** These columns are not required because they have too many null values in it and not so important information for analysis

```python
df['age'] = df['age'].fillna(df['age'].median())
df['fare'] = df['fare'].fillna(df['fare'].median())
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df= df.drop(columns= ['cabin','boat','body','home.dest'])
```

# Outlier Detection


```python
for col in ['age','fare','sibsp','parch']:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    IQR = q3-q1

    low = q1 - 1.5*IQR
    up = q3 + 1.5*IQR

    outlier = df[(df[col] < low) | (df[col] > up)]

    print(col)
    print('outlier:', len(outlier))
    print('-----------')
    print('low limit:', low)
    print('up limit:', up)
    print('-----------')
    
```

#### This is used to identify values in numerical columns. The detected outlier were retained because they can represent passenger information, rather than incorrect data.
